```python
df['at_bat'] = df['events'].map(at_bat_map)
df['is_swing'] = df['description'].map(swing_map)
df['contact'] = df['description'].map(contact_map)
df['batted_ball'] = df['description'].map(batted_ball_map)
df['in_zone'] = df['zone'].map(zone_map)
df['plate_appearance'] = df['events'].map(pa_map)
df['is_barrel'] = (df['launch_speed_angle'] == 6)
df['bb'] = df['events'].map(bb_map)
df['k'] = df['events'].map(k_map)
df['is_hard_hit'] = (df['launch_speed'] >= 95)
df['hbp'] = df['events'].map(hbp_map)
df['single'] = df['events'].map(b1_map)
df['double'] = df['events'].map(b2_map)
df['triple'] = df['events'].map(b3_map)
df['hr'] = df['events'].map(hr_map)
df['sf'] = df['events'].map(sf_map)
df['ibb'] = df['events'].map(ibb_map)
```

```python
o_swing = df.loc[df.in_zone == 0].groupby(['batter','season'], as_index = False).agg({'is_swing':'mean'})
hh_rate = df.loc[df.batted_ball == 1].groupby(['batter','season'], as_index = False).agg({'is_hard_hit':'mean'})
z_swing = df.loc[df.in_zone == 1].groupby(['batter','season'], as_index = False).agg({'is_swing':'mean'})
barrell = df.loc[df.batted_ball == 1].groupby(['batter','season'], as_index = False).agg({'is_barrel':'mean'})
at_bats = df.groupby(['batter','season'], as_index = False).agg({'at_bat':'sum'})
contact_rate = df.loc[df.is_swing == 1].groupby(['batter','season'], as_index = False).agg({'contact':'mean'})
k_rate = df.loc[df.plate_appearance == 1].groupby(['batter','season'], as_index = False).agg({'k':'mean'})
bb_rate = df.loc[df.plate_appearance == 1].groupby(['batter','season'], as_index = False).agg({'bb':'mean'})
cross_plate = df.groupby(['batter','season'], as_index = False).agg({'intercept_ball_minus_batter_pos_y_inches':'mean'})
```

```python
woba = df.groupby(['batter','season'], as_index = False).agg({'bb':'sum','hbp':'sum','single':'sum','double':'sum','triple':'sum','hr':'sum','at_bat':'sum',
                                                        'sf':'sum','ibb':'sum'})
woba['woba'] = (0.69*(woba.bb-woba.ibb) + 0.72*woba.hbp + 0.88*woba.single + 1.25*woba.double + 1.58*woba.triple + 2.01*woba.hr) / (woba.at_bat + woba.bb - woba.ibb + woba.sf + woba.hbp)
```

```python
df = o_swing.merge(z_swing, on = ['batter','season'])
df = df.merge(barrell, on = ['batter','season'])
df = df.merge(woba[['batter','season','woba']], on = ['batter','season'])
df = df.merge(at_bats, on = ['batter','season'])
df = df.merge(contact_rate, on = ['batter', 'season'])
df = df.merge(k_rate, on = ['batter', 'season'])
df = df.merge(bb_rate, on = ['batter', 'season'])
df = df.merge(hh_rate, on = ['batter', 'season'])
df = df.merge(cross_plate, on = ['batter', 'season'])

df.columns = ['batter','season','o-swing%','z-swing%','barrel%','woba','at_bats','contact%','K%','BB%','hard_hit%','intercept_ball_minus_batter_pos_y_inches']
```

```python
df['next_season'] = df['season'] + 1
```

```python
df = df.loc[df.at_bats >= df.at_bats.max()/3]
```

```python
df = df.merge(df[['batter','season','woba']], left_on = ['batter','next_season'], right_on = ['batter','season'], how = 'left')
```

# Modelling

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import pandas as pd
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error
```

```python
df1 = df.loc[df.season_x.isin([2023, 2024])]
df1 = df1.dropna()
df2 = df.loc[df.season_x == 2025]
```

```python
features = ['o-swing%', 'z-swing%', 'barrel%', 'contact%', 'hard_hit%', 'intercept_ball_minus_batter_pos_y_inches','woba_x']
```

```python
X = df1[features]
y = df1['woba_y']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.01,
    random_state=42,
    shuffle=True
)
```

```python
model = XGBRegressor(
    n_estimators=300,
    max_depth=8,
    learning_rate=0.01,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42
)

model.fit(X_train, y_train)
preds = model.predict(X_test)

rmse = mean_squared_error(y_test, preds) ** 0.5
print("RMSE:", rmse)
```

```python
preds_26 = model.predict(df2[features])
```

```python
df2['xWOBA_2026'] = preds_26
```
